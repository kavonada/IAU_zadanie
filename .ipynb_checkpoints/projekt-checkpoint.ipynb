## 1.1 C) Párová analýza dát: vzťahy a závislostí medzi dvojicami atribútov

Pred výpočtom korelácií sme vytvorili pracovnú kópiu, v ktorej sme dočasne nahradili zjavne chybné hodnoty hodnotou NaN, pretože extrémne hodnoty môžu skreslovať korelačný koeficient. Pôvodné dáta zostávajú nezmenené, ich čistenie budeme riešiť neskôr v časti 1.2 B).

```python
tmp = readings_data[sensor_cols].copy()
tmp.loc[tmp["oilTemperature"] < -900, "oilTemperature"] = np.nan
tmp.loc[tmp["motor_current"] > 20, "motor_current"] = np.nan
```

### Heatmapa korelácií (Pearson)
Heatmapa ukazuje lineárnu koreláciu pre všetky dvojice senzorov naraz.

```python
fig, ax = plt.subplots(figsize=(10, 8))
sns.heatmap(tmp.corr(), ax=ax, annot=True, fmt=".2f",
            cmap="coolwarm", center=0, vmin=-1, vmax=1)
ax.set_title("Pearsonova korelácia senzorov")
plt.xticks(rotation=45, ha="right")
plt.tight_layout()
plt.show()
```

Jediný silný vzťah je pressure_reservoir a pressureOil (r = 0,85). Sú silne kladne korelované, keď stúpne tlak v zásobníku, stúpne aj tlak oleja. Medzi ostatnými senzormi nie je žiadny lineárny vzťah, sú navzájom prakticky nezávislé.

Taktiež na zistenie presných hodnôt `r` si vytvoríme korelačnú maticu rovnakú ako pri Heatmape a vypíšeme najsilnejšie a najslabšie korelované dvojice

```python
corr = tmp.corr()
display(corr)

pairs = (corr.where(np.triu(np.ones(corr.shape), k=1).astype(bool))
             .stack().dropna().rename("r").reset_index())
pairs.columns = ["x", "y", "r"]
pairs["abs_r"] = pairs["r"].abs()
pairs = pairs.sort_values("abs_r", ascending=False).reset_index(drop=True)

print("Top 5 najsilnejších párov:")
display(pairs.head(5))
print("Top 5 najslabších párov:")
display(pairs.tail(5))
```

Okrem pressure_reservoir a pressureOil (r = 0,85) žiadne iné atribúty nemajú veľkú koreláciu.

### Scatter plot `pressure_reservoir` vs. `pressureOil`

```python
sns.regplot(data=readings_data, x="pressure_reservoir", y="pressureOil",
            scatter_kws={"alpha": 0.3}, line_kws={"color": "red"})

plt.title("pressure_reservoir vs pressureOil (r = 0,85)")
plt.xlabel("pressure_reservoir")
plt.ylabel("pressureOil")
plt.show()
```

Graf zobrazuje vzťah medzi `pressure_reservoir` a `pressureOil` (tlak v zásobníku vs tlak oleja). Každý bod je jedno meranie. Body sa sústreďujú okolo rastúcej priamky: keď stúpa tlak v zásobníku, stúpa aj tlak oleja. Tým pádom vieme povedať, že ide o kladný lineárny vzťah. To nám potvrdzuje predošlé tvrdenie z Pearsonovho koeficientu r = 0,85.

Vzťah je silný. Body tvoria pás okolo priamky, nie jednu čiaru, takže tlak oleja nie je jednoznačne určený tlakom v zásobníku. Nepozorujeme krivku, zhluky ani výrazné odľahlé body. Vzťah je z fyzikálneho hľadiska logický, lebo oba tlaky sa merajú v tom istom systéme a ovplyvňuje ich rovnaký stav. 

Taktiež sme skúšali Scatter ploty pre rôzne dvojice atribútov, no čiara bola vždy vodorovná bez žiadnych vzťahov. Boli testované aj Pairploty, ale boli ťažko čitateľné a dokazovali to isté ako Scatter plot.

## 1.1 D) Párová analýza dát: závislosti medzi predikovanou premennou a ostatnými premennými

Predikovaná premenná `Failure` je v súbore `readings`. Najprv skúmame jej rozdelenie a potom vzťah každého senzora k nej. Hodnota Failure môže byť buď 0 alebo 1 (binárna premenná). Preto používame boxploty, porovnanie priemerov a mediánov podľa tried a koreláciu s kódovanou premennou.

```python
print(readings_data["Failure"].value_counts(dropna=False))
print(readings_data["Failure"].value_counts(normalize=True, dropna=False).round(4))

readings_data["Failure"].value_counts().plot(kind="bar")
plt.title("Zastúpenie tried Failure")
plt.xlabel("Failure")
plt.ylabel("Počet záznamov")
plt.show()
```

Failure 0 (bez zlyhania) tvorí 12 249 záznamov (87,9 %).  
Failure 1 (zlyhanie) 1 690 záznamov (12,1 %).  
Dáta sú nevyvážené. Z tohto dôvodu pri modelovaní budeme používať metriky vhodné pre nevyvážené triedy, ako sú recall, precision, F1...

```python
tmp["Failure"] = readings_data["Failure"]     # znova pouzivam tmp, pretoze pri oilTemperature by vysiel nasledujuci rozdiel vyssi (cca o 0.6)
df_mean = tmp.groupby("Failure")[sensor_cols].mean().T
df_mean["rozdiel"] = df_mean[1] - df_mean[0]
display(df_mean)
```

Z tabuliek možno vidno ako sa menil priemer pri hodnote `Failure` 0 a 1. Pri atribútoch `oilTemperature`, `pressure_reservoir`, `pressureOil`, `rms_vibration`, `ambientNoise` a `aux_current` sú hodnoty v rozmedzí 0.0 - 0.3, `motor_current`, `humidityIntake` a `flowmeter` v rozmedzí 0.3 - 0.5, `dewpoint_discharge` v rozmedzí 0.5 - 1.0 a `filter_dp` výrazne najväčší rozdiel (až 3.64).

```python
fig, axes = plt.subplots(3, 4, figsize=(18, 11))
for ax, c in zip(axes.flat, sensor_cols):
    sns.boxplot(x="Failure", y=c, data=tmp, ax=ax)
    ax.set_title(f"{c} podľa Failure")
for ax in axes.flat[len(sensor_cols):]:
    ax.set_visible(False)
plt.tight_layout()
plt.show()
```

Boxy oboch tried sa pri všetkých senzoroch výrazne prekrývajú, žiadny senzor teda samostatne nerozdelí zlyhania od bežnej prevádzky. Pri `oilTemperature`, `flowmeter`, `humidityIntake` a `ambientNoise` sú rozdelenia v oboch triedach prakticky rovnaké. Čo sa týka outlierov, ich počet je výrazne väčší v triede 0 čo súvisí s ich väčším počtom záznamov (87,9 %).

```python
corr_f = pd.DataFrame({
    "Pearson": tmp[sensor_cols].corrwith(tmp["Failure"]),
}).round(3)
corr_f = corr_f.reindex(corr_f["Pearson"].abs().sort_values(ascending=False).index)
display(corr_f)

corr_f["Pearson"].plot(kind="barh", figsize=(8, 5))
plt.title("Korelácia senzorov s Failure (Pearson)")
plt.xlabel("r")
plt.show()
```

Z tabuľky a grafu možno vidieť, že všetky senzory majú so zlyhaním len slabý vzťah. Najviac sa líši `filter_dp`, ktorý je pri zlyhaní vyšší, tlaky sú naopak nižšie. Senzory `humidityIntake` a `ambientNoise` so zlyhaním nesúvisia vôbec.

## 1.2 A) Identifikácia problémov v dátach

Ako prvé skontrolujeme duplicitné záznamy

```python
units_dup = units_data.duplicated().sum()
readings_dup = readings_data.duplicated().sum()
gateways_dup = gateways_data.duplicated().sum()

print(f"Počet duplicitných hodnôt:")
print(f"units: {units_dup}")
print(f"readings: {readings_dup}")
print(f"gateways: {gateways_dup}")
```

Jediný súbor, ktorý obsahuje duplicitné hodnoty je readings.csv, konkrétne obsahuje 206 takých záznamov.

```python
units_nan = units_data.isna().any(axis=1).sum()
readings_nan = readings_data.isna().any(axis=1).sum()
gateways_nan = gateways_data.isna().any(axis=1).sum()

print(f"Počet riadkov s chýbajúcimi hodnotami:")
print(f"units: {units_nan}")
print(f"readings: {readings_nan}")
print(f"gateways: {gateways_nan}")
```

Všetky súbory obsahujú riadky s chýbajúcimi hodnotami, pričom `units.csv` ich obsahuje 100 v stĺpci hoursService, `readings.csv` 1725 v stĺpcoch rms_vibration, filter_dp a humidityIntake. `gateways.csv` ich obsahuje 14000 v stĺpci unit_id. V prvotnej fáze projektu ešte nebudeme chýbajúce hodnoty riešiť, pretože nevieme, či budú predstavovať problém pri riešení projektu. V dalších fázach môžu byť podľa potreby aplikované riešenia, ako napríklad dosadenie priemernej hodnoty z dát za chýbajúcu hodnotu, alebo odstránenie takých záznamov.

### Úprava datasetu units
Najprv zjednotíme formát dátumov na `datetime`

```python
units_data['last_overhaul'] = pd.to_datetime(units_data['last_overhaul'], format='mixed', errors='coerce')
```

Následne zjednotíme formát v stĺpci `dutyClass` pomocou mapovania hodnôt

```python
units_data['dutyClass'] = units_data['dutyClass'].astype(str).str.strip()

duty_class_map = {
    # Light (Trida 1)
    '1': 'light',
    'L': 'light',
    'light': 'light',
    
    # Medium (Trida 2 / Normal)
    '2': 'medium',
    'N': 'medium',
    'normal': 'medium',
    'med': 'medium',
    
    # Heavy (Trida 3)
    '3': 'heavy',
    'H': 'heavy',
    'heavy': 'heavy'
}

units_data['dutyClass'] = units_data['dutyClass'].map(duty_class_map)
units_data
```

### Úprava datasetu readings

Odstránime duplicitné záznamy

```python
readings_data = readings_data.drop_duplicates()
```

Zjednotíme formáty dátumov na `datetime` a id_unit na `Int64`

```python
readings_data['ts_cycle'] = pd.to_datetime(readings_data['ts_cycle'], format='mixed', errors='coerce')

readings_data['id_unit'] = (readings_data['id_unit'].astype(str).str.extract(r'(\d+)')[0].astype('Int64'))
readings_data
```

### Úprava datasetu gateways
Zjednotíme formáty dátumov na `datetime` a id_unit na `Int64`

```python
gateways_data['dateInstall'] = pd.to_datetime(gateways_data['dateInstall'], format='mixed', errors='coerce')
gateways_data['id_unit'] = (gateways_data['id_unit'].astype(str).str.extract(r'(\d+)')[0].astype('Int64'))
gateways_data
```

## 1.2) B Kontrola správnosti v dátach

V časti `1.1 B` sme identifikovali 2 problémové hodnoty, a to:

 - -999 v `oilTemperature`
 - 99 v `motor_current`

### Detekcia a odstránenie abnormálnych hodnôt

Teraz zistíme koľko takýchto záznamov sa v dátach nachádza

```python
abnormal_current = readings_data[readings_data['motor_current'] == 99]
print(f"Počet abnormálnych záznamov (motor_current): {len(abnormal_current)}")
abnormal_oil = readings_data[readings_data['oilTemperature'] == -999]
print(f"Počet abnormálnych záznamov (oilTemperature): {len(abnormal_oil)}")
```

Identifikovali sme 11 abnormálnych záznamov v stĺpci `motor_current` a 35 v stĺpci `oilTemperature`. Chybné záznamy nahradíme hodnotou NaN.

```python
readings_data['motor_current'] = readings_data['motor_current'].replace(99, np.nan)
readings_data['oilTemperature'] = readings_data['oilTemperature'].replace(-999, np.nan)
```

## 1.2) C Vychýlené hodnoty (outlier detection)

Ako prvé zpočítame, koľko záznamov z každého stĺpca predstavujú outliery a aký tvoria percentuálny podieľ z celkového počtu dát. Na základe toho rozhodneme, ktorú z dvoch techník následne použijeme pre ich odstránenie. Zvolili sme, že pre podiel outlierov menej než 1% Z celkového počtu záznamov outlierov úplne odstránime. Ak podiel outlierov bude predstavovať viac ako 1%, tak tieto hodnoty nahradíme hraničnými hodnotam z rozdelenia. Ak by sme aj tieto záznamy úplne odstránili, prišli by sme o príliš veľkú časť datasetu. 

```python
sensor_cols = [
    "motor_current", "filter_dp", "oilTemperature", "pressure_reservoir",
    "pressureOil", "rms_vibration", "flowmeter", "humidityIntake",
    "ambientNoise", "dewpoint_discharge", "aux_current"
]

outlier_results = []
total_rows = len(readings_data)

for col in sensor_cols:
    # 1. Výpočet kvartilov a IQR
    Q1 = readings_data[col].quantile(0.25)
    Q3 = readings_data[col].quantile(0.75)
    IQR = Q3 - Q1
    
    # 2. Určenie dolnej a hornej hranice
    lower_bound = Q1 - 1.5 * IQR
    upper_bound = Q3 + 1.5 * IQR
    
    # 3. Spočítanie outlierov
    outliers_count = ((readings_data[col] < lower_bound) | (readings_data[col] > upper_bound)).sum()
    
    # 4. Výpočet percentuálneho podielu zo všetkých riadkov
    outliers_pct = (outliers_count / total_rows) * 100
    
    outlier_results.append({
        'Senzor': col,
        'Počet outlierov': outliers_count,
        'Podiel (%)': round(outliers_pct, 2)
    })

# Vytvorenie prehľadného DataFrame
outliers_df = pd.DataFrame(outlier_results)
outliers_df
```

Najväčší počet outlierov sme detekovali v stĺpcoch `filter_dp` (539 alebo 3,92%) a  `rms_vibration` (487 alebo 3.55%). Vo všetkých ostatných stĺpcoch tvoria outliery menej ako 1%.

### Nahradenie vychýlenej hodnoty hraničnými hodnotami rozdelenia

Pre stĺpce s viac ako 1% vychýlenými záznammi nájdeme hraničné hodnoty na 5. a 95. percentile. Následne upravíme vychýlené hodnoty do rozsahu medzi tymito hodnotami.

```python
cols_to_clip = ['filter_dp', 'rms_vibration']

for col in cols_to_clip:
    # Vypočítame 5. a 95. percentil
    lower_percentile = readings_data[col].quantile(0.05)
    upper_percentile = readings_data[col].quantile(0.95)
    
    # Nahradíme hodnoty mimo tohto rozsahu
    readings_data[col] = readings_data[col].clip(lower=lower_percentile, upper=upper_percentile)
    
    print(f"Rozsah {col}: {lower_percentile:.4f} - {upper_percentile:.4f}")
```

Pre `filter_dp` predstavuje spodná hranica 11,47 a horná hranica 45,06. Pre `rms_vibration` spodná hranica predstavuje 1,18 a horná hranica 6,24.

### Odstránenie vychýlených alebo odľahlých pozorovaní

Pre stĺpce s podielom outlierov pod 1% záznamy odstránime a na konci spočítame celkový počet odstránených záznamov.

```python
cols_to_drop = [
    'oilTemperature', 'pressure_reservoir', 'pressureOil', 
    'flowmeter', 'ambientNoise', 'dewpoint_discharge', 'aux_current'
]

# Inicializujeme masku (všetko je False)
outlier_mask = pd.Series([False] * len(readings_data), index=readings_data.index)

for col in cols_to_drop:
    # Pre každý stĺpec vypočítame medze podľa IQR
    Q1 = readings_data[col].quantile(0.25)
    Q3 = readings_data[col].quantile(0.75)
    IQR = Q3 - Q1
    lower_bound = Q1 - 1.5 * IQR
    upper_bound = Q3 + 1.5 * IQR
    
    # Zistíme, ktoré riadky sú pre tento stĺpec outlieri
    col_outliers = (readings_data[col] < lower_bound) | (readings_data[col] > upper_bound)
    
    # Pridáme ich do celkovej masky
    outlier_mask = outlier_mask | col_outliers

print(f"Deleted sum: {outlier_mask.sum()}")

readings_data = readings_data[~outlier_mask]
```

Celkovo bolo odstránených 653 záznamov z vychýlenými hodnotami.

Nakoniec zobrazíme distribúciu dát po úpravách na grafoch.

```python
for sen in sensor_cols:
    fig, ax = plt.subplots(1, 2, figsize=(11, 3))
    readings_data[sen].plot.hist(bins=30, ax=ax[0], title=f"Histogram: {sen}")
    readings_data[sen].plot.box(vert=False, ax=ax[1], title=f"Boxplot: {sen}")
    plt.show()
```

Z vizualizácií je zreteľne vidieť úspešnú aplikáciu oboch zvolených metód na prácu s outliermi:

 - **Stĺpce ošetrené odstránením záznamov (napr. `oilTemperature`, `pressure_reservoir`):** Distribúcia týchto dát po očistení ukazuje krásne, takmer ukážkové normálne (zvonovité) rozdelenie. Ojedinelé body na koncoch boxplotov už nepredstavujú chybné merania, ale sú len prirodzeným štatistickým dôsledkom prepočítania užšieho medzikvartilového rozpätia (IQR) na novom, zomknutejšom datasete.

 - **Stĺpce ošetrené nahradením hraničnými hodnotami (napr. `filter_dp`):** Na histograme vidíme očakávaný vizuálny prejav nahradzovania hodnôt pomocou 5. a 95. percentilu. Hodnoty sú umelo nahromadené na oboch okrajoch distribúcie, čo vytvára výrazné stĺpce na minimálnej (cca 12) a maximálnej (cca 45) povolenej hranici. Boxplot pre tento senzor potvrdzuje, že za týmito "zastropovanými" hranicami sa už nenachádzajú žiadne extrémne odchýlky

 - **Celkový stav dát: Zvyšné stĺpce (napr. `motor_current`)** ukazujú kompaktné rozloženie bez prítomnosti akýchkoľvek anomálií za hranicami boxplotov.

# 1.3 A) Formulácia a štatistické overenie hypotézy

## 1.3 A) Hypotéza 1

`filter_dp` má v priemere vyššiu hodnotu v stave so zlyhaním ako bez zlyhania.

- **H0:** priemerná hodnota `filter_dp` nie je pri zlyhaní vyššia
- **H1:** priemerná hodnota `filter_dp` je pri zlyhaní vyššia
- **Hladina významnosti:** α = 0,05  

Test beží na očistených dátach: zástupné hodnoty (99, -999) sú nahradené NaN, záznamy s odľahlými hodnotami v ostatných senzoroch sú odstránené a `filter_dp` je orezaný na 5. a 95. percentil.

```python
tmp = readings_data[sensor_cols].copy()
tmp["Failure"] = readings_data["Failure"]
```

```python
zlyhanie = tmp[tmp["Failure"] == 1]["filter_dp"].dropna()
bez_zlyhania = tmp[tmp["Failure"] == 0]["filter_dp"].dropna()

print("So zlyhaním")
print("Počet:", len(zlyhanie))
print("Priemer:", round(zlyhanie.mean(), 2))
print("Medián:", round(zlyhanie.median(), 2))
print("Štandardná odchýlka:", round(zlyhanie.std(), 2))
print("")
print("Bez zlyhania")
print("Počet:", len(bez_zlyhania))
print("Priemer:", round(bez_zlyhania.mean(), 2))
print("Medián:", round(bez_zlyhania.median(), 2))
print("Štandardná odchýlka:", round(bez_zlyhania.std(), 2))
```

```python
print("Šikmosť so zlyhaním:", round(stats.skew(zlyhanie), 2))
print("Šikmosť bez zlyhania:", round(stats.skew(bez_zlyhania), 2))

sns.histplot(bez_zlyhania, stat="density", alpha=0.5, label="Bez zlyhania")
sns.histplot(zlyhanie, stat="density", alpha=0.5, label="So zlyhaním")
plt.title("Rozdelenie filter_dp podľa Failure")
plt.xlabel("filter_dp")
plt.legend()
plt.show()
```

Obe rozdelenia sú pravostranne zošikmené a nezodpovedajú normálnemu rozdeleniu. Pri takejto šikmosti nie je t-test vhodný, lebo predpokladá približne normálne dáta. Preto ako hlavný test použijeme Mann-Whitneyho U test, ktorý normalitu nevyžaduje. Normalitu overíme Anderson-Darlingovým testom.

```python
# Mann-Whitney
mw = stats.mannwhitneyu(zlyhanie, bez_zlyhania, alternative="greater")
print("Mann-Whitney test:", mw)
print("")

# Anderson-Darling
ad_zlyhanie = stats.anderson(zlyhanie, dist="norm")
ad_bez = stats.anderson(bez_zlyhania, dist="norm")

print("Anderson-Darling so zlyhaním, štatistika:", round(ad_zlyhanie.statistic, 2))
print("Anderson-Darling bez zlyhania, štatistika:", round(ad_bez.statistic, 2))
print("Kritické hodnoty:", ad_zlyhanie.critical_values)
print("Hladiny významnosti (%):", ad_zlyhanie.significance_level)

# t-test
t_test = stats.ttest_ind(zlyhanie, bez_zlyhania, equal_var=False, alternative="greater")
print("\nt-štatistika:", round(t_test.statistic, 2))
print("p-hodnota:", t_test.pvalue)
```

### Vyhodnotenie hypotézy 1

Normalitu sme pôvodne chceli overiť Shapiro-Wilkovým testom, no ten sa odporúča pre vzorky do 5 000 záznamov a skupina bez zlyhania ich má 11 263. Preto sme použili Anderson-Darlingov test, ktorý pracuje so všetkými záznamami.

Zároveň sme si rozdelili vzorku na `so zlyhaním` a `bez zlyhania`. Vzorka `so zlyhaním` má 1 275 záznamov pričom vzorka `bez zlyhania` 11 263. To či bude vzorka 1 275 záznamov dostatočne veľká zistíme pri ďalšom počítaní štatistickej sily. Priemer pri zlyhaní je 26,2 a bez zlyhania 23,1, medián 24,45 a 20,95.

Anderson-Darling test nám pomôže vybrať ten správny test. Štatistika vyšla 20,36 (so zlyhaním) a 263,37 (bez zlyhania), pričom kritická hodnota pri hladine významnosti 5 % je 0,752. Obe štatistiky ju výrazne prekračujú:  
- so zlyhaním: 20,36 > 0,752  
- bez zlyhania: 263,37 > 0,752
  
Normalitu v oboch skupinách zamietame, preto nepoužívame t-test, ale zvolíme Mann-Whitneyho test.

**Výsledok:** P-hodnota Mann-Whitneyho testu vyšla 2.345051972268802e-30, čo je veľmi malé číslo(0,00000...2). Preto vieme povedať, že p < 0,001, takže nulovú hypotézu zamietame a prijímame **H1:** priemerná hodnota `filter_dp` je pri zlyhaní významne vyššia.  

### 1.3 B) Veľkosť efektu a štatistická sila 

```python
from statsmodels.stats.power import TTestIndPower

n1 = len(zlyhanie)
n0 = len(bez_zlyhania)

spolocna_std = np.sqrt(((n1 - 1) * zlyhanie.var() + (n0 - 1) * bez_zlyhania.var()) / (n1 + n0 - 2))
d_efekt = (zlyhanie.mean() - bez_zlyhania.mean()) / spolocna_std
print("Cohenovo d:", round(d_efekt, 3))

analyza = TTestIndPower()
sila = analyza.power(effect_size=d_efekt, nobs1=n1, ratio=n0 / n1, alpha=0.05, alternative="larger")
print("Štatistická sila:", round(sila, 3))
```

### Zhodnotenie pre hypotézu 1

Štatistickú silu testu zistíme využitím triedy `TTestIndPower` z knižnice `statsmodels`.

**Výsledky:**
- Cohenovo d = 0,336
- Odhadovaná štatistická sila = 1,0 (100 %)

Cohenovo d vyšlo 0,336, čo je malý až stredný efekt. Rozdiel priemerov `filter_dp` medzi skupinami je teda približne tretina smerodajnej odchýlky. Štatistická sila vyšla 1,0, čo je výrazne nad odporúčanou hranicou 0,8. Ak teda rozdiel medzi skupinami naozaj existuje, test ho odhalí takmer s istotou. Preto dokážeme povedať, že dáta poskytujú dostatočnú podporu pre túto hypotézu.

**Záver hypotézy 1**  
`filter_dp` je pri zlyhaní štatisticky významne vyšší. Rozdiel je však malý až stredný a rozdelenia oboch skupín sa výrazne prekrývajú, takže `filter_dp` samotný zlyhanie nepredikuje.

## 1.3 A) Hypotéza 2

`pressure_reservoir` má v priemere nižšiu hodnotu v stave so zlyhaním ako bez zlyhania.

* **H0:** priemerná hodnota `pressure_reservoir` nie je pri zlyhaní nižšia.
* **H1:** priemerná hodnota `pressure_reservoir` je pri zlyhaní nižšia.
* **Hladina významnosti:** α = 0,05

Pri testovaní budeme pracovať s vytvorenou kópiou dát `tmp` (očistenou od chýbajúcich a extrémnych hodnôt).

```python
# Rozdelenie dát pre atribút pressure_reservoir (s odstránením chýbajúcich hodnôt)
zlyhanie_pres = tmp[tmp["Failure"] == 1]["pressure_reservoir"].dropna()
bez_zlyhania_pres = tmp[tmp["Failure"] == 0]["pressure_reservoir"].dropna()

print("So zlyhaním")
print("Počet:", len(zlyhanie_pres))
print("Priemer:", round(zlyhanie_pres.mean(), 2))
print("Medián:", round(zlyhanie_pres.median(), 2))
print("Štandardná odchýlka:", round(zlyhanie_pres.std(), 2))
print("")
print("Bez zlyhania")
print("Počet:", len(bez_zlyhania_pres))
print("Priemer:", round(bez_zlyhania_pres.mean(), 2))
print("Medián:", round(bez_zlyhania_pres.median(), 2))
print("Štandardná odchýlka:", round(bez_zlyhania_pres.std(), 2))
```

Z prvotného prehľadu vidíme, že `pressure_reservoir` je pri zlyhaní o niečo nižší `8,59` v porovnaní so stavom bez zlyhania `8,72`. Smerodajné odchýlky sú takmer identické, čo naznačuje podobnú variabilitu v oboch skupinách.

```python
print("Šikmosť so zlyhaním:", round(stats.skew(zlyhanie_pres), 2))
print("Šikmosť bez zlyhania:", round(stats.skew(bez_zlyhania_pres), 2))

sns.histplot(bez_zlyhania_pres, stat="density", alpha=0.5, label="Bez zlyhania")
sns.histplot(zlyhanie_pres, stat="density", alpha=0.5, label="So zlyhaním")
plt.title("Rozdelenie pressure_reservoir podľa Failure")
plt.xlabel("pressure_reservoir")
plt.legend()
plt.show()
```

Histogramy aj vypočítaná šikmosť (`-0,00` a `-0,02`) ukazujú, že hodnoty v oboch skupinách majú veľmi symetrické rozdelenie, ktoré sa vizuálne javí ako normálne. Graf takisto potvrdzuje, že hoci je stred hodnôt pri zlyhaní posunutý mierne doľava (k nižším hodnotám), obe rozdelenia sa výrazne prekrývajú.

```python
# Anderson-Darling test pre normalitu
ad_zlyhanie_pres = stats.anderson(zlyhanie_pres, dist="norm")
ad_bez_pres = stats.anderson(bez_zlyhania_pres, dist="norm")

print("Anderson-Darling so zlyhaním, štatistika:", round(ad_zlyhanie_pres.statistic, 2))
print("Anderson-Darling bez zlyhania, štatistika:", round(ad_bez_pres.statistic, 2))
print("Kritické hodnoty:", ad_zlyhanie_pres.critical_values)
print("Hladiny významnosti (%):", ad_zlyhanie_pres.significance_level)
print("")
t2 = stats.ttest_ind(zlyhanie_pres, bez_zlyhania_pres, equal_var=False, alternative="less")
print("Welch t-test: t =", round(t2.statistic, 2),", p =", t2.pvalue)
```

Keďže Anderson-Darlingov test zamietol normalitu pre skupinu bez zlyhania (štatistika `2,39 > 0,752`), opierame sa o neparametrický Mann-Whitneyho test.

```python
# Mann-Whitney U test
mw_pres = stats.mannwhitneyu(zlyhanie_pres, bez_zlyhania_pres, alternative="less")
print("Mann-Whitney test:", mw_pres)
print("")
```

Mann-Whitneyho test dal p < 0,001 (3.78670923531442e-18), takže hodnoty `pressure_reservoir` sú pri zlyhaní štatisticky významne nižšie.

### Vyhodnotenie hypotézy 2

Na základe Anderson-Darlingovho testu (štatistika `0.53` pre zlyhanie a `2.39` bez zlyhania) zisťujeme, že dáta pri zlyhaní spĺňajú podmienku normality (hodnota je nižšia ako kritická hranica 0,752), zatiaľ čo dáta bez zlyhania normalitu mierne porušujú. Celkovo je však rozdelenie vizuálne veľmi blízke k normálnemu, čo potvrdzuje aj takmer nulová šikmosť (`-0,0` so zlyhaním a `-0,02` bez zlyhania).

**Výsledok:** Vzhľadom na mierne porušenie normality v jednej skupine sme sa opreli o robustnejší Mann-Whitneyho test, ktorý však potvrdil výsledky Welchovho t-testu. Oba testy vyšli s p-hodnotou p < 0,001, čo znamená, že nulovú hypotézu **zamietame** a prijímame alternatívnu hypotézu **H1**. Priemerná hodnota `pressure_reservoir` pri zlyhaní `8,59` je štatisticky významne nižšia ako bez zlyhania `8,72`. 

### 1.3 B) Veľkosť efektu a štatistická sila 

```python
n1_pres = len(zlyhanie_pres)
n0_pres = len(bez_zlyhania_pres)

# Výpočet Cohenovho d
spolocna_std_pres = np.sqrt(((n1_pres - 1) * zlyhanie_pres.var() + (n0_pres - 1) * bez_zlyhania_pres.var()) / (n1_pres + n0_pres - 2))
d_efekt_pres = (zlyhanie_pres.mean() - bez_zlyhania_pres.mean()) / spolocna_std_pres
print("Cohenovo d (pôvodné):", round(d_efekt_pres, 3))
print("Cohenovo d (absolútna veľkosť efektu):", round(abs(d_efekt_pres), 3))

# Výpočet štatistickej sily (s použitím absolútnej hodnoty efektu)
analyza_pres = TTestIndPower()
sila_pres = analyza_pres.power(effect_size=abs(d_efekt_pres), nobs1=n1_pres, ratio=n0_pres / n1_pres, alpha=0.05, alternative="larger")
print("Štatistická sila:", round(sila_pres, 3))
```

Štatistická sila testu je 1,0 (100 %). To znamená, že ak rozdiel s touto veľkosťou efektu naozaj existuje, test ho odhalí takmer s istotou. Absolútna hodnota Cohenovho d `(0,24)` však indikuje len malý efekt, čo znamená, že rozdiel v tlaku je síce merateľný, ale z hľadiska praktickej využiteľnosti je relatívne slabý.

### Zhodnotenie pre hypotézu 2

* Absolútna hodnota Cohenovho d = `0,24`
* Odhadovaná štatistická sila = `1,0 (100 %)`

Cohenovo d vyšlo v absolútnej hodnote `0,24`, čo predstavuje malý efekt. Rozdiel medzi skupinami je teda na úrovni približne štvrtiny smerodajnej odchýlky (ktorá je okolo `0,58`). To znamená, že atribút `pressure_reservoir` ako samostatný izolovaný indikátor zlyhania nestačí.

## 2.1 A) Rozdelenie na trénovaciu a testovaciu množinu

Pred rozdelením spojíme tri súbory do jednej tabuľky, kde jeden riadok predstavuje jedno meranie. Vychádzame z dát očistených vo fáze 1 (zjednotené formáty, odstránené duplicity, zástupné hodnoty nahradené `NaN`, ošetrené odľahlé hodnoty). Spájame `readings`, `units`, `gateways` cez `id_unit`. Brány bez priradenej jednotky (14 000 zo 16 000) sa k žiadnemu meraniu pripojiť nedajú, preto ich pred spojením vynecháme. Merania, ktorých `id_unit` neexistuje v `units`, vyradíme, lebo by nemali žiadne informácie o jednotke ani bráne.

```python
def merge_tables(readings, units, gateways, verbose=True):
    n_readings = len(readings)

    # 1) merania, ku ktorým neexistuje jednotka v units
    unmatched_mask = ~readings["id_unit"].isin(units["id_unit"])
    unmatched = readings.loc[unmatched_mask]

    # 2) brány bez priradenej jednotky sa k žiadnemu meraniu pripojiť nedajú
    gw_linked = gateways.dropna(subset=["id_unit"])

    # 3) spojenie len pre priradené merania; validate="m:1" zhodí kód, ak by sa riadky množili
    merged = (readings.loc[~unmatched_mask]
              .merge(units, on="id_unit", how="left", validate="m:1")
              .merge(gw_linked, on="id_unit", how="left", validate="m:1", suffixes=("", "_gw")))

    assert len(merged) == n_readings - len(unmatched), "spojenie zmenilo počet riadkov"

    # depotId je v units aj gateways; ak sa zhoduje, stačí jeden stĺpec
    same = (merged["depotId_gw"].isna() | (merged["depotId"] == merged["depotId_gw"])).all()
    if same:
        merged = merged.drop(columns="depotId_gw")

    if verbose:
        n_no_gw = merged["id_bramy_check"].isna().sum() if "id_bramy_check" in merged else merged["id_gateway"].isna().sum()
        print(f"Meraní pred spojením:                {n_readings}")
        print(f"Nepriradených (id_unit nie je v units): {len(unmatched)} "
              f"({len(unmatched) / n_readings * 100:.2f} %), unikátnych ID: {unmatched['id_unit'].nunique()}")
        if len(unmatched):
            print(f"   rozsah ID: {unmatched['id_unit'].min()} – {unmatched['id_unit'].max()}")
            if "Failure" in unmatched:
                print(f"   podiel Failure=1 medzi nimi: {unmatched['Failure'].mean():.3f} "
                      f"(v celých dátach: {readings['Failure'].mean():.3f})")
        print(f"Meraní po spojení:                   {len(merged)}")
        print(f"Meraní bez brány v gateways:         {n_no_gw}")
        print(f"depotId v units a gateways sa zhoduje: {same}")
    return merged, unmatched


data_merged, unmatched = merge_tables(readings_data, units_data, gateways_data)
data_merged
```

Z 13 080 meraní (po čistení vo fáze 1) sa 134 (1,02 %, 120 unikátnych ID) nedalo priradiť k žiadnej jednotke. Všetky majú ID v rozsahu 90001 – 90498, rovnako ako vo fáze 1. Po spojení má tabuľka 12 946 riadkov. Každé meranie má priradenú bránu a `depotId` sa v `units` aj `gateways` zhoduje, preto ponechávame len jeden stĺpec.

Dáta rozdelíme na trénovaciu a testovaciu množinu v pomere 80 : 20. Rozdelenie je stratifikované podľa `Failure`, pretože zlyhania tvoria len približne 12 % dát a bez stratifikácie by sa ich podiel v množinách mohol líšiť. Fixný `random_state` zabezpečí, že rozdelenie bude pri každom spustení rovnaké. Od tohto bodu pracujeme len s trénovacou množinou.

```python
import os
from sklearn.model_selection import train_test_split

RANDOM_STATE = 42 # seed na replikovanie rovnakych vysledkov
TARGET = "Failure"
DATA_DIR = "data"
OUT_DIR = "output"

os.makedirs(OUT_DIR, exist_ok=True)

train_df, test_df = train_test_split(data_merged, test_size=0.2, stratify=data_merged[TARGET], random_state=RANDOM_STATE)

print("train:", train_df.shape, "| test:", test_df.shape)
print("podiel Failure=1  train: %.3f | test: %.3f" % (train_df[TARGET].mean(), test_df[TARGET].mean()))
print("jednotiek v oboch množinách naraz:", len(set(train_df["id_unit"]) & set(test_df["id_unit"])),
      "z", data_merged["id_unit"].nunique())
```

Trénovacia množina má 10 356 a testovacia 2 590 meraní. Podiel zlyhaní je v oboch 12,2 %, čo potvrdzuje správnu stratifikáciu. Každá jednotka má v dátach viac meraní, preto sa 1 455 z 1 996 jednotiek nachádza v oboch množinách. Keďže pozorovaním je jedno meranie, ponechávame náhodné rozdelenie. Vo fáze 3 však treba počítať s tým, že model mohol vidieť merania tej istej jednotky v trénovacej množine, čo môže výsledky na teste mierne nadhodnotiť.

## 2.1 B) Transformácia dát na vhodný formát

Prekonvertujeme formát dátumov, aby bola splnená požiadavka všetkých atribútov v numerickom formáte. Nový formát vyjadruje počet uplynutých dní od referenčného dátumu 1970-01-01.

```python
EPOCH = pd.Timestamp("1970-01-01")
DATE_COLS = ["ts_cycle", "last_overhaul", "dateInstall"]


def dates_to_numeric(d):
    d = d.copy()
    for c in DATE_COLS:
        d[c + "_num"] = (d[c] - EPOCH).dt.total_seconds() / 86400
    return d


train_df = dates_to_numeric(train_df)
test_df = dates_to_numeric(test_df)
train_df[[c + "_num" for c in DATE_COLS]].describe().round(1)
```

Z trénovacej a testovacej množiny vyradíme náš TARGET, stĺpce opisujúce identifikátory a pôvodné dátumy.

```python
ID_COLS = ["id_unit", "numberSerial", "macAddress", "Ipv4", "id_gateway"]

y_train = train_df[TARGET].astype(int)
X_train = train_df.drop(columns=[TARGET] + ID_COLS + DATE_COLS)

print(X_train.shape)
X_train.dtypes.to_frame("dtype").join(X_train.nunique().rename("počet rôznych hodnôt"))
```

Po vyradení cieľovej premennej, identifikátorov (`id_unit`, `numberSerial`, `macAddress`, `Ipv4`, `id_gateway`) a pôvodných dátumov zostalo 27 atribútov: 19 číselných a 8 textových. Identifikátory majú unikátne alebo takmer unikátne hodnoty, nenesú zovšeobecniteľnú informáciu a model by si ich mohol len zapamätať. Z textových atribútov majú `modelCode` (4), `version_firmware` (3), `hardwareModel` (5) a `dutyClass` (3) málo hodnôt, kým `install_region` (60), `owner_operator` (40), `vendor` (40) a `technician` (120) majú vysokú kardinalitu. `depotId` je síce číselný, ale ide o identifikátor depa (120 hodnôt), nie o meranú veličinu.

Záznamy ktoré nie sú v numerickom formáte, ale sú dôležité pre naše riešenie pretypujeme na numerický formát a to následovne:

- `dutyClass`: light -> 0, medium -> 1, heavy -> 2
- nízka kardinalita (`modelCode`, `version_firmware`, `hardwareModel`): použijeme one-hot kódovanie. Každá hodnota atribútu sa stane samostatným binárnym stĺpcom, ktorý má hodnotu 1, ak má dané meranie túto hodnotu, a 0 v opačnom prípade.
- vysoká kardinalita (`install_region`, `owner_operator`, `vendor`, `technician`, `depotId`): vyradíme. One-hot kódovanie by pridalo približne 380 riedkych stĺpcov a ide o administratívne údaje, ktoré neopisujú technický stav jednotky.

```python
from sklearn.compose import ColumnTransformer
from sklearn.preprocessing import OneHotEncoder, OrdinalEncoder

MAX_CARDINALITY = 15
ORDINAL_COLS = ["dutyClass"]
DUTY_ORDER = ["light", "medium", "heavy"]

nominal = [c for c in X_train.select_dtypes(exclude="number").columns if c not in ORDINAL_COLS]
card = X_train[nominal].nunique()
OHE_COLS = card[card <= MAX_CARDINALITY].index.tolist()
HIGH_CARD_COLS = card[card > MAX_CARDINALITY].index.tolist() + ["depotId"]   # depotId je ID, nie veličina
NUM_COLS = [c for c in X_train.select_dtypes(include="number").columns if c not in HIGH_CARD_COLS]

print("one-hot:", OHE_COLS)
print("vyradené (vysoká kardinalita):", HIGH_CARD_COLS)

encoder = ColumnTransformer(
    [("num", "passthrough", NUM_COLS),
     ("duty", OrdinalEncoder(categories=[DUTY_ORDER]), ORDINAL_COLS),
     ("ohe", OneHotEncoder(handle_unknown="ignore", sparse_output=False), OHE_COLS)],
    remainder="drop", verbose_feature_names_out=False)
encoder.set_output(transform="pandas")

X_train_num = encoder.fit_transform(X_train)     # fit len na train
X_train_num.head()
```

Overenie toho, či každé pozorovanie predstavuje jeden riadok, a či sú všetky atribúty numerické

```python
assert len(X_train_num) == len(y_train)                       # 1 pozorovanie = 1 riadok
assert X_train_num.index.is_unique
assert X_train_num.select_dtypes(exclude="number").empty      # všetky atribúty sú číselné
print("OK:", X_train_num.shape, "| chýbajúce hodnoty:", int(X_train_num.isna().sum().sum()))
```

## Transformery

Pre každý šikmý atribút porovná pôvodné hodnoty s tromi transformáciami: Yeo-Johnson, Box-Cox a Quantile (na normálne rozdelenie)

```python
SKEWED_COLS = ["hoursService", "filter_dp", "rms_vibration"]

def transform_variants(s):
    x = s.dropna().to_numpy().reshape(-1, 1)
    return {
        "pôvodné": x,
        "Yeo-Johnson": PowerTransformer(method="yeo-johnson").fit_transform(x),
        "Box-Cox": PowerTransformer(method="box-cox").fit_transform(x),
        "Quantile -> normal": QuantileTransformer(output_distribution="normal",
                                                  random_state=RANDOM_STATE).fit_transform(x),
    }


skew_rows = {}
for col in SKEWED_COLS:
    v = transform_variants(X_train[col])
    skew_rows[col] = {name: round(float(stats.skew(a.ravel())), 2) for name, a in v.items()}
    fig, axes = plt.subplots(1, len(v), figsize=(3.2 * len(v), 2.8))
    for ax, (name, a) in zip(axes, v.items()):
        ax.hist(a.ravel(), bins=30)
        ax.set_title(name, fontsize=9)
    fig.suptitle(col)
    plt.tight_layout()
    plt.show()

print("Šikmosť po transformácii:")
pd.DataFrame(skew_rows).T
```

Použili sme Yeo-Johnsonovu transformáciu, ktorá znížila šikmosť na približne 0. Box-Cox dáva rovnaký výsledok, ale vyžaduje striktne kladné hodnoty, preto Yeo-Johnson považujeme za bezpečnejšiu voľbu. QuantileTransformer sme zamietli, lebo hodnoty zhluknuté na orezaných hraniciach (`filter_dp`, `rms_vibration`) premietol na extrémy +-5 a pri `hoursService` šikmosť znížil len na 1,08.

## Scalery

Porovnáme dva scalery: `StandardScaler` (z-normalizácia), `MinMaxScaler` (normalizácia do rozsahu 0–1) . Ukážku robíme na dvoch senzoroch, ktoré majú rôzny charakter rozdelenia hodnôt: `pressure_reservoir` a `humidityIntake`. Pre každý scaler sledujeme priemer, smerodajnú odchýlku, minimum a maximum po transformácii.

```python
demo_cols = ["pressure_reservoir", "humidityIntake"]
scalers = {"StandardScaler": StandardScaler(), "MinMaxScaler": MinMaxScaler()}

rows = []
for col in demo_cols:
    x = X_train[[col]].dropna()
    rows.append({"atribút": col, "scaler": "pôvodné", "priemer": x[col].mean(), "std": x[col].std(),
                 "min": x[col].min(), "max": x[col].max()})
    for name, sc in scalers.items():
        z = pd.Series(sc.fit_transform(x).ravel())
        rows.append({"atribút": col, "scaler": name, "priemer": z.mean(), "std": z.std(),
                     "min": z.min(), "max": z.max()})
pd.DataFrame(rows).round(3)
```

`pressure_reservoir` má približne normálne rozdelenie a `humidityIntake` rovnomerné v rozsahu 10–95. `StandardScaler` dal obom priemer 0 a smerodajnú odchýlku 1. `MinMaxScaler` ich previedol do rozsahu 0–1. Pri rovnomernom `humidityIntake` je výsledná smerodajná odchýlka vyššia (0.289) než pri `pressure_reservoir` (0.178). Z toho vyplýva rozdelenie v pipeline: `StandardScaler` pre približne normálne senzory a `MinMaxScaler` pre rovnomerne rozložené a ohraničené atribúty (`humidityIntake`, `lat`, `lon`, číselné dátumy).

## Výsledná transformácia

Na základe pozorovaní z experimentov sme si rozdelili dáta do 5 skupín:

- *(SKEWED_COLS):* stĺpce so šikmým rozdelením, imputácia mediánom, transformácia `Yeo-Johnson` metódou so štandardizáiou. 
- *STD_COLS:* stĺpce s približne normálnym rozdelením, imputácia mediánom, škálovanie pomocou `StandardScaler`.
- *MINMAX_COLS:* rovnomerné alebo ohraničené stĺpce, imputácia mediánom, škálovanie pomocou `MinMaxScaler`
- *ORDINAL_COLS:* ordinálny stĺpce (len `dutyClass`)
- *OHE_COLS:* nominále stĺpce, použitý `OneHotEncoder`


```python
SKEWED_COLS = ["hoursService", "filter_dp", "rms_vibration"]
STD_COLS = ["motor_current", "oilTemperature", "pressure_reservoir", "pressureOil",
            "flowmeter", "ambientNoise", "dewpoint_discharge", "aux_current", "signal_dbm"]
MINMAX_COLS = ["humidityIntake", "lat", "lon", "ts_cycle_num", "last_overhaul_num", "dateInstall_num"]
ORDINAL_COLS = ["dutyClass"]
OHE_COLS = ["modelCode", "version_firmware", "hardwareModel"]

def make_preprocess(skewed=SKEWED_COLS, std=STD_COLS, minmax=MINMAX_COLS):
    return ColumnTransformer([
        ("skewed", Pipeline([("imp", SimpleImputer(strategy="median")),
                             ("power", PowerTransformer(method="yeo-johnson"))]), skewed),
        ("std", Pipeline([("imp", SimpleImputer(strategy="median")),
                          ("scale", StandardScaler())]), std),
        ("minmax", Pipeline([("imp", SimpleImputer(strategy="median")),
                             ("scale", MinMaxScaler())]), minmax),
        ("duty", OrdinalEncoder(categories=[DUTY_ORDER]), ORDINAL_COLS),
        ("ohe", OneHotEncoder(handle_unknown="ignore", sparse_output=False), OHE_COLS),
    ], verbose_feature_names_out=False).set_output(transform="pandas")


preprocess = make_preprocess()
X_train_prep = preprocess.fit_transform(X_train)
X_train_prep.join(y_train).to_csv(f"{OUT_DIR}/train_preprocessed.csv", index=False)
display(X_train_prep.head())
print(X_train_prep.shape)
```

Výsledná trénovacia množina má 10356 riadkov a 31 atribútov, z toho 12 pribudlo z one-hot kódovania. Kontroly zo zadania sú splnené: jedno pozorovanie je opísané jedným riadkom, všetky atribúty sú číselné a v dátach nezostali žiadne chýbajúce hodnoty. Výsledok je uložený v `output/train_preprocessed.csv`.

`preprocess` je objekt `ColumnTransformer`, ktorý je naučený len na trénovacích dátach, takže ho možno bez zmeny použiť na testovaciu množinu cez `preprocess.transform(X_test)` (replikovateľnosť, 2.3). Testovaciu množinu sme v tejto fáze nepoužili.

- **Rozdelenie:** Dáta sme rozdelili v pomere 80 : 20, stratifikovane podľa `Failure` a s fixným `random_state`, takže podiel zlyhaní je v oboch množinách 12,2 % a rozdelenie je zopakovateľné. Rozdelenie predchádza všetkým krokom fázy 2, ktoré sa učia z dát (imputácia, transformácie, škálovanie).
- **Dátumy:** Dátumy sme previedli na počet dní od 1. 1. 1970, čo je pevná konštanta a nespôsobuje únik informácií. Rozdiely dátumov sme neodvodzovali, lebo pri ~14 % nekonzistentných meraní by vznikali nelogické záporné hodnoty.
- **Identifikátory:** Vyradili sme `id_unit`, `numberSerial`, `macAddress`, `Ipv4` a `id_gateway`, pretože majú unikátne hodnoty a model by si ich len zapamätal. Nenesú žiadnu zovšeobecniteľnú informáciu o stave jednotky.
- **Kódovanie kategórií:** `dutyClass` kódujeme ordinálne (light = 0, medium = 1, heavy = 2), lebo kategórie majú prirodzené poradie podľa záťaže. `modelCode`, `version_firmware` a `hardwareModel` kódujeme one-hot (12 stĺpcov), pretože majú málo hodnôt a žiadne poradie.
- **Vysoká kardinalita:** Atribúty `install_region`, `owner_operator`, `vendor`, `technician` a `depotId` majú 40 – 120 hodnôt a one-hot kódovanie by pridalo približne 380 riedkych stĺpcov. Ide o administratívne údaje, ktoré priamo neopisujú technický stav jednotky, preto sme ich vyradili.
- **Chýbajúce hodnoty:** Chýbajúce hodnoty dopĺňame mediánom, ktorý je odolný voči odľahlým hodnotám a pri šikmých atribútoch lepšie zodpovedá typickej hodnote než priemer. Imputáciu pomocou susedov (KNN) sme nepoužili, lebo senzory medzi sebou prakticky nekorelujú (fáza 1).
- **Transformer:** Na `hoursService`, `filter_dp` a `rms_vibration` sme použili Yeo-Johnsonovu transformáciu, ktorá znížila šikmosť z 1,52 / 0,83 / 0,78 na približne 0. Box-Cox dal rovnaký výsledok, ale vyžaduje kladné hodnoty, a QuantileTransformer vytvoril umelé extrémy ±5 z hodnôt zhluknutých na orezaných hraniciach.
- **Scalery:** Približne normálne senzory škálujeme pomocou `StandardScaler` a rovnomerné či ohraničené atribúty (`humidityIntake`, `lat`, `lon`, číselné dátumy) pomocou `MinMaxScaler`. Voľbu sme urobili podľa tvaru rozdelenia, ktorý sme overili porovnaním na `pressure_reservoir` a `humidityIntake`.
- 
Výsledná trénovacia množina má 10 356 riadkov a 31 číselných atribútov bez chýbajúcich hodnôt a je uložená v `output/train_preprocessed.csv`.

### Známe obmedzenia

- **Odľahlé hodnoty sú ošetrené pred rozdelením.** Hranice IQR a percentily z fázy 1 sme vypočítali na celých dátach, takže zahŕňajú aj testovacie riadky, a odstránených 653 riadkov (~4,8 %) chýba aj v testovacej časti. Metodicky správnejšie by bolo tieto hranice učiť len z trénovacej množiny. Vplyv považujeme za malý, ide o vedomé zjednodušenie.
- **Orezanie na percentily** spôsobilo, že ~10 % hodnôt `filter_dp` a `rms_vibration` leží presne na hraniciach. Yeo-Johnson tieto zhluky neodstráni, preto rozdelenie nie je zvonovité, hoci šikmosť je ~0.
- **Merania tej istej jednotky** sú v trénovacej aj testovacej množine (1 455 z 1 996 jednotiek), čo môže výsledky na teste mierne nadhodnotiť.

## 2.2 A)  3 techniky výberu atribútov

```python
from sklearn.feature_selection import mutual_info_classif, RFE
from sklearn.ensemble import RandomForestClassifier
```

Pre lepšiu predstavu s čím budeme pracovať, si vypíšeme ako také numerické riadky vyzerajú.

```python
display(X_train_num.head(3))
```

```python
X = X_train_prep.copy() 
y = y_train.loc[X.index]

print("Počet záznamov:", X.shape[0], "| atribútov:", X.shape[1], "| chýbajúce hodnoty:", int(X.isna().sum().sum()))
```

Pracujeme s výstupom 2.1 C (X_train_prep), ktorý je už imputovaný, transformovaný a zakódovaný. Trénovacie dáta sú v premennej X a hodnoty Failure v premennej y.

### 1. Technika: Mutual Information

Mutual Information je technika, ktorá meria závislosti jednotlivých atribútov s danou premennou `Failure`. Čím vyššie skóre, tým je atribút dôežitejší, pretože má väčšiu spojitosť s cieľovou premennou.

```python
# Identifikácia zakódovaných kategóriových atribútov
ohe_names = preprocess.named_transformers_["ohe"].get_feature_names_out(OHE_COLS)
discrete_mask = X.columns.isin(["dutyClass", *ohe_names])

mi_scores = mutual_info_classif(
    X,
    y,
    discrete_features=discrete_mask,
    n_neighbors=3,
    random_state=42
)

mi_results = pd.Series(
    mi_scores,
    index=X.columns,
    name="Mutual Information"
).sort_values(ascending=False)

display(mi_results.to_frame())
```

Najvyššie skóre dosiahol znova atribút `filter_dp` a `rms_vibration` zo senzorových meraní. No naviac pribudli s relatívne dobrým skóre aj atribúty z iných súborov ako sú `hoursService`, `last_overhaul_num`, `lat`, `lon` a `dutyClass`. Pri rôznych experimentoch meniť premennú `n_neighbors` sme dostávali podobné výsledky, a preto nastavujeme hodnotu na 3.

Keďže Mutual Information hodnotí každý atribút samostatne a nezohľadňuje ich vzájomné kombinácie, výsledky ďalej porovnáme s inými technikami.

### 2. Technika: Random Forest

Ako druhú techniku výberu atribútov sme použili RFE (Recursive Feature Elimination) s modelom Random Forest. Ide o techniku trénovania modelu, pričom postupne vyhadzujeme najmenej dôležité atribúty, až kým nezostane požadovaný počet atribútov.

```python
rf = RandomForestClassifier(
    n_estimators=100,
    class_weight="balanced",
    random_state=42,
    n_jobs=-1
)

rfe = RFE(
    estimator=rf,
    n_features_to_select=5,
    step=0.2
)

rfe.fit(X, y)

rfe_results = pd.DataFrame({
    "Atribut": X.columns,
    "RFE poradie": rfe.ranking_,
}).sort_values("RFE poradie")

display(rfe_results)
```

V modeli sme zvolili `n_features_to_select=5` pre zjednodušenie modelu a pre finálne porovnanie top 5 atribútov všetkých troch techník.

Najdôležitejšími atribútmi podľa metódy RFE s modelom Random Forest sa ukázali `filter_dp`, `rms_vibration`, `motor_current`, `hoursService` a `aux_current`, ktoré dosiahli poradie 1. S porovnaním s Mutual Information sa opäť potvrdila dôležitosť atribútov `filter_dp` a `rms_vibration`, ale RFE navyše uprednostnila `motor_current` a `aux_current`. Naopak, atribúty `lat`, `lon` a `last_overhaul_num`, ktoré mali pri Mutual Information relatívne dobré skóre, RFE vyradila oveľa skôr.

### 3. Technika: Logistická regresia s L1 regularizáciou

Ako tretiu techniku výberu atribútov sme použili logistickú regresiu s L1 regularizáciou. Táto technika počas trénovania modelu priraďuje atribútom koeficienty, pričom menej dôležitým atribútom môže nastaviť koeficient na 0. Čím vyššia je absolútna hodnota koeficientu, tým väčší vplyv má atribút na rozhodovanie modelu.

Pred L1 dáta navyše štandardizujeme, aby boli koeficienty porovnateľné, lebo výstup 2.1 C obsahuje stĺpce v rôznych mierkach (z-skóre, rozsah 0–1, binárne a ordinálne hodnoty).

```python
# Škálovanie dát pre logistickú regresiu
scaler = StandardScaler()
X_scaled = scaler.fit_transform(X)

# Logistická regresia s L1 regularizáciou
logreg_l1 = LogisticRegression(
    l1_ratio=1.0,
    solver="liblinear",
    C=0.05,
    class_weight="balanced",
    max_iter=3000,
    random_state=42
)

logreg_l1.fit(X_scaled, y)

# Absolútne hodnoty koeficientov
l1_scores = np.abs(logreg_l1.coef_[0])

l1_results = pd.DataFrame({
    "Atribut": X.columns,
    "L1 koeficient": l1_scores,
}).sort_values("L1 koeficient", ascending=False)

display(l1_results)
```

Najvyššie koeficienty dosiahli atribúty `dutyClass`, `motor_current`, `hoursService`, `filter_dp` a `aux_current`. Oproti predchádzajúcim technikám tu najviac vynikol atribút `dutyClass`. Opäť sa však potvrdila dôležitosť atribútov `filter_dp`, `hoursService` a `motor_current`.

Naopak, atribúty ako `lon` a `signal_dbm` dostali nulové koeficienty, takže ich model pri tomto nastavení regularizácie nevyužíva.

Silu regularizácie sme nastavili pomocou parametra `C=0.05`. Menšie hodnoty C znamenajú silnejšiu regularizáciu, ktorá môže vynulovať viac koeficientov. Taktiež sme experimentovali s rôznymi hodnotami parametra C, konkrétne 0.01, 0.03 a 0.08.

```python
# Porovnanie techník
comparison = pd.DataFrame({
    "Atribut": X.columns,
    "MI skore": mi_scores,
    "RFE poradie": rfe.ranking_,
    "L1 koeficient": l1_scores
})

# TOP 5 atribútov
mi_top5 = set(comparison.nlargest(5, "MI skore")["Atribut"])
rfe_top5 = set(X.columns[rfe.support_])
l1_top5 = set(comparison.nlargest(5, "L1 koeficient")["Atribut"])

comparison["MI TOP 5"] = comparison["Atribut"].isin(mi_top5)
comparison["RFE TOP 5"] = comparison["Atribut"].isin(rfe_top5)
comparison["L1 TOP 5"] = comparison["Atribut"].isin(l1_top5)

comparison["Pocet vyberov"] = (
    comparison[["MI TOP 5", "RFE TOP 5", "L1 TOP 5"]]
    .sum(axis=1)
)

# Zoradenie
comparison = comparison.sort_values(
    ["Pocet vyberov", "MI skore"],
    ascending=[False, False]
)

comparison = comparison[comparison["Pocet vyberov"] > 0]

display(comparison)

# Graf
comparison.set_index("Atribut")["Pocet vyberov"].sort_values().plot(
    kind="barh",
    figsize=(10, 3)
)

plt.title("Porovnanie techník výberu atribútov")
plt.xlabel("Počet techník, ktoré vybrali atribút do TOP 5")
plt.xticks([0, 1, 2, 3])
plt.tight_layout()
plt.show()
```

### Porovnanie a vyhodnotenie techník

Na výber atribútov sme použili tri rôzne techniky: **Mutual Information, RFE s Random Forest a logistickú regresiu s L1 regularizáciou**. Výsledky sme porovnali podľa piatich najlepších atribútov z každej techniky.

**Prečo zrovna tieto 3 techniky?**

Vybrali sme si tieto tri techniky, pretože každá pristupuje k výberu atribútov iným spôsobom. V prednáške boli opísané 3 rôzne prístupy: **Filter, Wrapper a Embedded**. Rozhodli sme sa použiť jednu techniku z každého prístupu, aby sme mohli porovnať ich výsledky

- Mutual Information (Filter) – hodnotí každý atribút samostatne.

- RFE s Random Forest (Wrapper) – opakovane trénuje model a postupne odstraňuje najmenej dôležité atribúty.

- Logistická regresia s L1 regularizáciou (Embedded) – počas trénovania modelu znižuje koeficienty menej užitočných atribútov.


Jediný atribút, ktorý sa umiestnil medzi piatimi najlepšími vo všetkých troch technikách, je `filter_dp`. V dvoch technikách sa zhodli `rms_vibration`, `hoursService`, `motor_current` a `aux_current`. Atribút `hoursService` budeme brať s rezervou, pretože môže obsahovať informácie z budúcnosti, čo predstavuje riziko úniku informácií (data leakage). 

Jednotlivé techniky sa však v niektorých výsledkoch líšili. Napríklad atribút `dutyClass` dosiahol najvyšší koeficient pri L1 regularizácii, zatiaľ čo pri ostatných technikách sa medzi piatimi najlepšími neumiestnil.

Na základe porovnania považujeme za najperspektívnejšie atribúty `filter_dp`, `rms_vibration`, `motor_current` a `aux_current`.

Tieto atribúty odporúčame ďalej overiť pri trénovaní a vyhodnocovaní modelov, pretože samotné metódy výberu atribútov ešte nezaručujú najlepšiu predikčnú úspešnosť.

## 2.2 B)  Zoradenie atribútov podľa dôležitosti

```python
ranking = pd.DataFrame({
    "Atribut": X.columns,
    "MI skore": mi_scores,
    "RFE poradie": rfe.ranking_,
    "L1 koeficient": l1_scores
})

# Prevedieme výsledky na poradia (1 = najlepší)
ranking["MI poradie"] = ranking["MI skore"].rank(
    ascending=False, method="min"
)

ranking["L1 poradie"] = ranking["L1 koeficient"].rank(
    ascending=False, method="min"
)

ranking["Priemer poradi"] = ranking[
    ["MI poradie", "RFE poradie", "L1 poradie"]
].mean(axis=1)

# Zoradíme
ranking = ranking.sort_values(
    ["Priemer poradi", "MI poradie", "Atribut"],
    ascending=[True, True, True]
).reset_index(drop=True)

ranking.index = ranking.index + 1
ranking.index.name = "Celkove poradie"

display(ranking[
    ["Atribut", "MI poradie", "RFE poradie",
     "L1 poradie", "Priemer poradi"]
])
```

## 2.3 A) Replikovateľnosť predspracovania 

V tejto časti spájame predspracovanie dát z 2.1 a výber atribútov z 2.2 do jednej `Pipeline`. Pri `fit` sa všetky naučené kroky  určia iba z trénovacích dát. Pri `transform` testovacích dát použijeme už naučené transformácie.

Nadväzujeme na `train_df` a `test_df` z časti 2.1, kde už máme upravené dátumy a rozdelené dáta na trénovaciu a testovaciu množinu.

Predspracovanie preberáme z 2.1 C cez funkciu make_preprocess, takže pipeline používa rovnaké skupiny atribútov aj rovnaké transformácie. Tak isto aj zoznamy stĺpcov (`NUM_COLS`, `ORDINAL_COLS`, `OHE_COLS`) preberáme z 2.1. V `TARGET` máme uloženú hodnotu "Failure".

```python
from sklearn.feature_selection import SelectFromModel
from sklearn.compose import ColumnTransformer
```

```python
def split_X_y(df):
    # rovnaká úprava pre train aj test: vyradenie cieľa, identifikátorov a pôvodných dátumov
    X = df.drop(columns=[TARGET] + ID_COLS + DATE_COLS)
    y = df[TARGET].astype(int)
    return X, y


X_train_23, y_train_23 = split_X_y(train_df)
X_test_23, y_test_23 = split_X_y(test_df)

print("Train:", X_train_23.shape)
print("Test:", X_test_23.shape)
```

```python
EXCLUDED_23 = ["hoursService"]          # riziko úniku informácií (viď 2.2 B)


def without(cols):
    return [c for c in cols if c not in EXCLUDED_23]


# 5 najlepších atribútov zo spoločného poradia v 2.2 B, bez vyradených
SELECTED_23 = [a for a in ranking["Atribut"] if a not in EXCLUDED_23][:5]
print("Vybrané atribúty:", SELECTED_23)

pipeline_23 = Pipeline([
    # rovnaké predspracovanie ako v 2.1 C (imputácia, Yeo-Johnson, scalery, kódovanie)
    ("preprocessing", make_preprocess(skewed=without(SKEWED_COLS),
                                      std=without(STD_COLS),
                                      minmax=without(MINMAX_COLS))),
    # výber atribútov podľa 2.2
    ("feature_selection", ColumnTransformer([("sel", "passthrough", SELECTED_23)],
                                            verbose_feature_names_out=False)),
]).set_output(transform="pandas")
```

Pipeline má dva kroky. Prvý je predspracovanie z 2.1 C (imputácia mediánom, Yeo-Johnson, StandardScaler, MinMaxScaler, ordinálne a one-hot kódovanie). Druhý je výber piatich najlepších atribútov zo spoločného poradia v 2.2 B. hoursService vynechávame pre riziko úniku informácií.

```python
pipeline_23.fit(X_train_23, y_train_23)              # učí sa len z trénovacích dát

X_train_final_23 = pipeline_23.transform(X_train_23)
X_test_final_23 = pipeline_23.transform(X_test_23)    # test sa len transformuje

print("Train:", X_train_final_23.shape)
print("Test:", X_test_final_23.shape)

X_train_final_23.join(y_train_23).to_csv(f"{OUT_DIR}/train_final.csv", index=False)
X_test_final_23.join(y_test_23).to_csv(f"{OUT_DIR}/test_final.csv", index=False)
```

`fit` vykonáme výhradne na tréningovej množine. Testovaciu množinu spracujeme iba pomocou `transform`, aby z nej neunikali informácie do dopĺňania, kódovania alebo výberu atribútov.

```python
display(X_train_final_23.head())
```

Skontrolujeme, že výstupy nemajú chýbajúce hodnoty, obidve množiny majú rovnaké stĺpce a opakované `transform` na rovnakých dátach dáva identický výsledok.


```python
assert X_train_final_23.shape[1] == X_test_final_23.shape[1] == len(SELECTED_23)
assert X_train_final_23.columns.equals(X_test_final_23.columns)
assert not X_train_final_23.isna().any().any()
assert not X_test_final_23.isna().any().any()

# opakovaná transformácia dáva identický výsledok
np.testing.assert_allclose(X_test_final_23.to_numpy(), pipeline_23.transform(X_test_23).to_numpy())

# pipeline dáva na train rovnaký výsledok ako predspracovanie z 2.1 C
np.testing.assert_allclose(X_train_final_23.to_numpy(), X_train_prep[SELECTED_23].to_numpy())

print("Pipeline funguje správne.")
```
